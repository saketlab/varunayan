# Climate and public interest

This notebook compares ERA5 climate data with Google Trends search interest from 2004 to 2024. It has three examples: umbrella searches and rainfall in India, sunscreen searches and temperature in California, and dengue and malaria searches and rainfall in India. The Google Trends series are monthly CSV files from [Google Trends](https://trends.google.com/). Sales data for a product can replace them.

```python
import pandas as pd

import varunayan as v
```

## Rainfall in India

Download monthly total precipitation (mm) for India at 0.25°. The [LGD state boundaries](https://geo.bharatviz.org/geojsons/admin/India-geodata-lgd-states.geojson) set the region. Then take the area-weighted mean for each month.

```python
india = v.era5ify_geojson(
    request_id="tp_india_2004_2024",
    variables=["total_precipitation"],
    start_date="2004-01-01",
    end_date="2024-12-31",
    json_file="https://geo.bharatviz.org/geojsons/admin/India-geodata-lgd-states.geojson",
    frequency="monthly",
    resolution=0.25,
    verbosity=0,
)
rain = v.spatial_mean(india, "value", by=["year", "month"]).rename(
    columns={"value": "rainfall_mm"}
)
rain.head()
```

The map shows the mean June to August rainfall total for each grid cell.

```python
jja = v.seasonal_stat(india, "value", months=[6, 7, 8], how="sum")
v.plot_map(
    jja,
    "value",
    log=True,
    boundaries=v.load_india_boundaries(),
    legend_title="June-August rainfall (mm)",
)
```

## Umbrella searches and rainfall in India

Load the search interest for "umbrella" in India and merge it with the rainfall series.

```python
umbrella = pd.read_csv(
    "https://gist.githubusercontent.com/JaggeryArray/490818f9cd19336d63f9942589a61bf0/raw/92273fc7c39c5dcc71acb27242a2dd3d6b72b4ed/multiTimeline_india_googletrends_umbrella_2004-now.csv",
    parse_dates=["Month"],
).rename(columns={"Month": "date"})
umbrella_rain = rain.merge(umbrella, on="date")
v.plot_dual_axis(umbrella_rain, "date", "rainfall_mm", "umbrella", log_right=True)
```

Calculate the cross-correlation. Rainfall is `x` and search interest is `y`. A peak at a negative lag means that rainfall changes first.

```python
v.plot_ccf(v.cross_correlation(umbrella_rain["rainfall_mm"], umbrella_rain["umbrella"]))
```

> Both series have one peak in each year. The largest correlation is at lag +1 (r = 0.45), and r = 0.43 at lag 0. A positive lag means that search interest changes first. Umbrella searches increase approximately one month before the rainfall peak, because people prepare for the monsoon before it arrives.

## Sunscreen searches and temperature in California

Download monthly 2 m temperature (°C) for [California](https://gist.githubusercontent.com/JaggeryArray/50ad17645a290ee4445e1113609de5e4/raw/91e97ec6ae6654093490df87a9802e359bf303b1/california.geojson) and take the area-weighted mean for each month.

```python
california = v.era5ify_geojson(
    request_id="t2m_california_2004_2024",
    variables=["2m_temperature"],
    start_date="2004-01-01",
    end_date="2024-12-31",
    json_file="https://gist.githubusercontent.com/JaggeryArray/50ad17645a290ee4445e1113609de5e4/raw/91e97ec6ae6654093490df87a9802e359bf303b1/california.geojson",
    frequency="monthly",
    resolution=0.25,
    verbosity=0,
)
temp = v.spatial_mean(california, "value", by=["year", "month"]).rename(
    columns={"value": "t2m_c"}
)
temp.head()
```

Load the search interest for "sunscreen" in California and merge it with the temperature series.

```python
sunscreen = pd.read_csv(
    "https://gist.githubusercontent.com/JaggeryArray/05a2ffb206ae6ada04f5187c15d77d45/raw/99b5dac0efed5c3fa56c0221e2352aa2e358e17f/multiTimeline_california_googletrends_sunscreen_2004-now.csv",
    parse_dates=["Month"],
).rename(columns={"Month": "date"})
sunscreen_temp = temp.merge(sunscreen, on="date")
v.plot_dual_axis(sunscreen_temp, "date", "t2m_c", "sunscreen", log_right=True)
```

Temperature is `x` and search interest is `y`.

```python
v.plot_ccf(v.cross_correlation(sunscreen_temp["t2m_c"], sunscreen_temp["sunscreen"]))
```

> The largest correlation is at lag +1 (r = 0.48). Sunscreen searches increase approximately one month before the temperature peak. Search interest also increases over the 20 years, but the temperature does not show the same increase.

## Dengue and malaria searches and rainfall in India

Load the search interest for "dengue symptoms" and "malaria symptoms" in India. Merge both with the same rainfall series.

```python
dengue = pd.read_csv(
    "https://gist.githubusercontent.com/JaggeryArray/fcc204d10f0b7acdd4813f1581ab18ef/raw/3589dea8be071f545b94f236db1b52f2efd11723/multiTimeline_india_googletrends_dengue_symptoms_2004-now.csv",
    parse_dates=["Month"],
).rename(columns={"Month": "date"})
malaria = pd.read_csv(
    "https://gist.githubusercontent.com/JaggeryArray/5635a5654a0264d71b031fdd073cbf83/raw/8f8c760840507583d1b72a846ed76e3cdbc697ee/multiTimeline_india_googletrends_malaria_symptoms_2004-now.csv",
    parse_dates=["Month"],
).rename(columns={"Month": "date"})
disease_rain = rain.merge(dengue, on="date").merge(malaria, on="date")
v.plot_dual_axis(disease_rain, "date", "rainfall_mm", "dengue symptoms", log_right=True)
```

```python
v.plot_dual_axis(
    disease_rain, "date", "rainfall_mm", "malaria symptoms", log_right=True
)
```

Rainfall is `x` and search interest is `y`.

```python
v.plot_ccf(
    v.cross_correlation(disease_rain["rainfall_mm"], disease_rain["dengue symptoms"]),
    title="Dengue symptoms",
)
```

```python
v.plot_ccf(
    v.cross_correlation(disease_rain["rainfall_mm"], disease_rain["malaria symptoms"]),
    title="Malaria symptoms",
)
```

> For dengue, the largest correlation is at lag -2 and at lag +10 (r = 0.65 for both). The cycle repeats every 12 months, so the two lags show the same relation. Rainfall comes approximately two months before the peak in dengue searches. For malaria, the largest correlation is at lag -1 (r = 0.51). Rainfall comes approximately one month before the peak in malaria searches.

## Conclusion

- Umbrella searches in India and sunscreen searches in California increase approximately one month before the rainfall or temperature peak.
- Dengue and malaria searches in India increase one to two months after the rainfall peak. This agrees with the time that mosquitoes need to breed after the rain.

The same steps work for other regions, variables and search terms or sales data.
